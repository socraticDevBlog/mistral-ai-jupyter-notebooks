# 00 — Hello Mistral

Smoke test for the Mistral AI API: list available models, run a chat completion, and stream a response.

Setup: copy `.env.example` to `.env` at the project root and set `MISTRAL_API_KEY` (or export it in your shell).

```python
from mistral_jupyter import DEFAULT_MODEL, MAX_TOKENS, complete, get_client, stream

client = get_client()
print("client ready — default model:", DEFAULT_MODEL, "| max_tokens cap:", MAX_TOKENS)
```

## List available models

```python
models = client.models.list()
for m in models.data:
    print(m.id)
```

## Chat completion

`complete()` applies the project `MAX_TOKENS` cap by default.

```python
response = complete(
    [
        {"role": "system", "content": "You are a concise assistant."},
        {"role": "user", "content": "What is a proof-of-concept, in one sentence?"},
    ],
    client=client,
)

print(response.choices[0].message.content)
print("usage:", response.usage)
```

## Streaming

```python
for delta in stream(
    [{"role": "user", "content": "Count from 1 to 5, one number per line."}],
    client=client,
):
    print(delta, end="")
```
